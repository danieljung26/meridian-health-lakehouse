```sql
CREATE OR REPLACE FUNCTION meridian_dev.silver.rls_by_state(state STRING)
RETURN
  current_user() = 'ADMIN_PLACEHOLDER'   -- stand-in for the full-PHI role: sees all rows
  OR state = 'Massachusetts';             -- everyone else: only MA patients
```

```sql
ALTER TABLE meridian_dev.bronze.patients
  SET ROW FILTER meridian_dev.silver.rls_by_state ON (STATE);
```

```sql
SELECT STATE, count(*) AS n
FROM meridian_dev.bronze.patients
GROUP BY STATE
ORDER BY n DESC;
```
