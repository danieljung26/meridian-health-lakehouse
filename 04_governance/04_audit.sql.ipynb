```sql
SELECT *
FROM system.access.audit
WHERE action_name IN ('getTable', 'commandSubmit')
ORDER BY event_time DESC
LIMIT 20;
```

```sql
SELECT
    event_time,
    user_identity.email        AS user_email,
    action_name,
    request_params
FROM system.access.audit
WHERE event_date >= current_date() - INTERVAL 7 DAYS
  AND (
        lower(request_params.full_name_arg) LIKE '%patients%'
        OR lower(CAST(request_params AS STRING)) LIKE '%patients%'
      )
ORDER BY event_time DESC
LIMIT 50;
```
